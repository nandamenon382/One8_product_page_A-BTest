```python
import pandas as pd

df = pd.read_csv("one8_ab_test_sessions.csv")
```

```python
df[["variant", "revenue"]].head()
```

```python
from scipy import stats

rev_a = df[df.variant == "A"].revenue
rev_b = df[df.variant == "B"].revenue

t_stat, p_rev = stats.ttest_ind(
    rev_b,
    rev_a,
    equal_var=False
)
```

```python
print("===== REVENUE A/B TEST =====")
print(f"A mean revenue: ₹{rev_a.mean():.2f}")
print(f"B mean revenue: ₹{rev_b.mean():.2f}")
print(f"Revenue difference: ₹{rev_b.mean() - rev_a.mean():.2f}")
print(f"t-statistic: {t_stat:.3f}")
print(f"p-value: {p_rev:.4f}")
```

Variant B has a higher observed conversion rate than A. The two-proportion z-test produces a p-value of 0.00684, indicating that the difference is statistically significant at the 5% level.
